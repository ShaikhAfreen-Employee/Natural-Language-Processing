```python
#--------------------Definition-----------------------

# Text classification is the process of assigning text to a category.
# Here, we use SVM (Support Vector Machine) to classify movie reviews as Positive or Negative.

# What we are doing here;
# Text → TF-IDF (convert text to numbers) → SVM → Prediction.
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC

# Text data
texts = [
    "This movie was amazing",
    "I loved this film",
    "The movie was fantastic",
    "It was a wonderful experience",
    "The film was boring",
    "I hated this movie",
    "This movie was terrible",
    "The film was very disappointing"
]

# Labels
labels = [
    "Positive",
    "Positive",
    "Positive",
    "Positive",
    "Negative",
    "Negative",
    "Negative",
    "Negative"
]

# Convert text into TF-IDF features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)

# Create SVM model
model = SVC(kernel="linear")

# Train the model
model.fit(X, labels)

# Predict all texts
predictions = model.predict(X)

# Display output for every text
for text, prediction in zip(texts, predictions):
    print("Text:", text)
    print("Prediction:", prediction)
    print("-" * 40)
```
