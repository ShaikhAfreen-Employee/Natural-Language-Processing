```python
# Hierarchical Clustering is an unsupervised learning technique
# used to group similar data points into clusters.
# It creates a hierarchy of clusters represented using a dendrogram.

# It has two types:
# 1. Agglomerative Clustering - Bottom-up approach that merges clusters.
# 2. Divisive Clustering - Top-down approach that splits clusters.
```

```python
# -----------------Agglomerative Hierarchical Clustering in NLP-----------------------

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import pairwise_distances

# Sample text documents
documents = [
    "Machine learning is a branch of artificial intelligence",
    "Deep learning uses neural networks",
    "Artificial intelligence is used in robotics",
    "Football is a popular outdoor sport",
    "Cricket is played with a bat and ball",
    "Sports require physical fitness"
]

# Convert text into TF-IDF vectors
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(documents)

# Calculate cosine distance
distance_matrix = pairwise_distances(X.toarray(), metric='cosine')

# Agglomerative clustering
model = AgglomerativeClustering(
    n_clusters=2,
    metric='precomputed',
    linkage='average'
)

labels = model.fit_predict(distance_matrix)

# Display results
for i, document in enumerate(documents):
    print(f"Document {i+1}: Cluster {labels[i]}")
    print(document)
    print()
```

```python
# here, the algorithm groups AI/ML-related documents together and sports-related documents together(merging)
```

```python
# ----------------Divisive Hierarchical Clustering in NLP--------------------------

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Sample text documents
documents = [
    "Machine learning is a branch of artificial intelligence",
    "Deep learning uses neural networks",
    "Artificial intelligence is used in robotics",
    "Football is a popular outdoor sport",
    "Cricket is played with a bat and ball",
    "Sports require physical fitness"
]

# Convert documents into TF-IDF vectors
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(documents)

# First split: divide all documents into 2 clusters
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Display results
for i, document in enumerate(documents):
    print(f"Document {i+1}: Cluster {labels[i]}")
    print(document)
    print()
```

```python
# Divisive Hierarchical Clustering in NLP

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Sample text documents
documents = [
    "Machine learning is a branch of artificial intelligence",
    "Deep learning uses neural networks",
    "Artificial intelligence is used in robotics",
    "Football is a popular outdoor sport",
    "Cricket is played with a bat and ball",
    "Sports require physical fitness"
]

# Convert text documents into TF-IDF vectors
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(documents)


# Divisive clustering function
def divisive_clustering(X, indices, final_clusters):

    # Stop if cluster contains only one document
    if len(indices) <= 1:
        final_clusters.append(indices)
        return

    # Split the cluster into two sub-clusters
    kmeans = KMeans(
        n_clusters=2,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X[indices])

    # Create two sub-clusters
    cluster1 = [
        indices[i]
        for i in range(len(indices))
        if labels[i] == 0
    ]

    cluster2 = [
        indices[i]
        for i in range(len(indices))
        if labels[i] == 1
    ]

    # Recursively divide both clusters
    divisive_clustering(X, cluster1, final_clusters)
    divisive_clustering(X, cluster2, final_clusters)


# Start with all documents in one cluster
indices = list(range(len(documents)))

# Store final clusters
final_clusters = []

# Perform divisive clustering
divisive_clustering(X, indices, final_clusters)


for cluster_no, cluster in enumerate(final_clusters, start=1):

    print(f"\nCluster {cluster_no}:")

    for index in cluster:
        print(f"Document {index + 1}: {documents[index]}")
```

```python
# Divisive clustering starts with all documents in one cluster and recursively splits them into smaller clusters.
# TF-IDF represents the documents numerically, while K-Means performs the splitting based on document similarity.
```
