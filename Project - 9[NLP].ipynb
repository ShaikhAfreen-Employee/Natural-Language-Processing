```python
# ------------------Laplace Smoothing----------------------
# Laplace Smoothing, also called Add-One Smoothing, is a technique used in probability and machine learning ,
# to avoid zero probability for events or words that have not appeared in the training data.
# IN OTHER WORDS => * Laplace Smoothing is a technique that adds 1 to each word count ,to prevent zero probability for unseen words *.
```

```python
# Laplace Smoothing

def calculate_probability(word, word_counts, total_words, vocabulary_size):
    count = word_counts.get(word, 0)
    
    # Add-1 smoothing
    probability = (count + 1) / (total_words + vocabulary_size)
    
    return probability


# Training word counts
word_counts = {
    "Apple": 3,
    "Banana": 2,
    "Orange": 1
}

# Calculate values
total_words = sum(word_counts.values())
vocabulary_size = len(word_counts)

# Test words
test_words = ["Apple", "Banana", "Grape", "Orange", "Kiwi"]

for word in test_words:
    probability = calculate_probability(
        word,
        word_counts,
        total_words,
        vocabulary_size
    )
    
    print(word, ":", round(probability, 4))
```

```python
# Here,grape and kiwi was not present in the training data, but it still gets: 0.1 instead of 0......this is called laplace smoothing technique
```
