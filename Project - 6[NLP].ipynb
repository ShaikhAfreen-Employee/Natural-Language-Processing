```python
# Text Preprocessing
text = "The cat is sitting on the mat"

words = text.lower().split()

print("1. Tokenized Words:")
print(words)
```

```python
# Creating vocabulary
vocabulary = list(dict.fromkeys(words))

print("2. Vocabulary:")
print(vocabulary)
```

```python
# Load Pre-trained GloVe
print("3. Loading GloVe model...")

glove = api.load("glove-wiki-gigaword-50")

print("GloVe model loaded successfully!")
```

```python
# Get Word Vectors
print("4. Word Vectors:")

for word in vocabulary:
    if word in glove:
        print(word, "=", glove[word][:5], "...")
```

```python
# Performing Dot Product
word1 = "cat"
word2 = "mat"

dot_product = np.dot(glove[word1], glove[word2])

print("5. Dot Product:")
print("cat and mat =", round(dot_product, 4))
```

```python
# Finding Similarity
similarity = glove.similarity(word1, word2)

print("6. Similarity:")
print("Similarity between 'cat' and 'mat' =", round(similarity, 4))
```

```python
# Creating Embedding Matrix
embedding_matrix = np.array([
    glove[word] for word in vocabulary
])

print("7. Embedding Matrix Shape:")
print(embedding_matrix.shape)

print("\nEmbedding Matrix:")
print(embedding_matrix)
```
