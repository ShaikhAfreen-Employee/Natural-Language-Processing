```python
from gensim.models import Word2Vec
import gensim
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
import zipfile
import warnings

nltk.download('punkt')
warnings.filterwarnings(action='ignore')
```

```python
with zipfile.ZipFile("Gutenburg.zip", 'r') as zip_ref:
    file_name = zip_ref.namelist()[0]

    with zip_ref.open(file_name) as file:
        content = file.read().decode('utf-8', errors='ignore')

cleaned_text = content.replace("\n", " ")
print("File loaded")
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize

data = []

for sentence in sent_tokenize(cleaned_text):
    temp = []

    for word in word_tokenize(sentence):
        temp.append(word.lower())

    data.append(temp)

print("Tokenization completed")
```

```python
model1 = gensim.models.Word2Vec(
    data,
    min_count=1,
    vector_size=100,
    window=5,
    sg=0
)
```

```python
model2 = gensim.models.Word2Vec(
    data,
    min_count=1,
    vector_size=100,
    window=5,
    sg=1
)
```

```python
print("Cosine similarity between 'alice' and 'wonderland' (CBOW):",
      model1.wv.similarity('alice', 'wonderland'))

print("Cosine similarity between 'alice' and 'machines' (CBOW):",
      model1.wv.similarity('alice', 'machines'))
```
